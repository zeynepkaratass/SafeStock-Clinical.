```python
import pandas as pd
import random
from datetime import datetime, timedelta

items = [
    ("Nitrile Gloves, Size M", "Consumables", False),
    ("Surgical Masks", "Consumables", False),
    ("Syringe 5mL", "Consumables", False),
    ("Syringe 10mL", "Consumables", False),
    ("IV Cannula 20G", "Consumables", False),
    ("Gauze Sponges 4x4", "Consumables", False),
    ("Lidocaine 1%", "Medication", True),
    ("Lidocaine 2%", "Medication", True),
    ("Epinephrine 1:1000", "Medication", True),
    ("Ephedrine 50mg", "Medication", True),
    ("Saline Solution 0.9% 500ml", "Fluids", False),
    ("Dextrose 5% 500ml", "Fluids", False),
    ("Suture Silk 3-0", "Surgical", False),
    ("Scalpel Blades #10", "Surgical", False),
    ("Alcohol Prep Pads", "Consumables", False),
    ("Bandages 2 inch", "Consumables", False),
    ("Dopamine 400mg", "Medication", True),
    ("Dobutamine 250mg", "Medication", True),
    ("Insulin Regular", "Medication", True),
    ("Insulin Glargine", "Medication", True)
]

data = []
for i in range(1, 101):
    base_item = items[i % len(items)]
    item_name = f"{base_item[0]} - Batch {random.randint(100, 999)}"
    category = base_item[1]
    is_lasa = base_item[2]

    current_stock = random.randint(5, 500)
    minimum_required = random.randint(20, 100)
    unit_price = round(random.uniform(0.5, 50.0), 2)

    days_to_expire = random.randint(30, 730)
    exp_date = (datetime.now() + timedelta(days=days_to_expire)).strftime('%Y-%m-%d')

    data.append({
        "Item_ID": f"ITM-{1000+i}",
        "Item_Name": item_name,
        "Category": category,
        "Current_Stock": current_stock,
        "Minimum_Required": minimum_required,
        "Unit_Price_USD": unit_price,
        "Expiration_Date": exp_date,
        "LASA_Warning": is_lasa
    })

df = pd.DataFrame(data)
df.to_csv("clinic_inventory.csv", index=False)
print("100 verilik 'clinic_inventory.csv' file created successfully!")
```

```python
import pandas as pd

# Step 1: Initialize Database
try:
    df = pd.read_csv("clinic_inventory.csv")
    print("--- Clinical Inventory System Initialized ---")
except FileNotFoundError:
    print("Error: 'clinic_inventory.csv' not found. Please create the file first.")
    exit()

# Main Loop
while True:
    print("\n" + "="*40)
    # Step 2: Search Input
    item_to_find = input("Search item name (or 'Q' to quit): ").strip()

    if item_to_find.upper() == "Q":
        print("System closing... Have a safe shift!")
        break

    # Step 3: Search Operation
    results = df[df['Item_Name'].str.contains(item_to_find, case=False, na=False)]

    if not results.empty:
        print("\n--- Matching Items Found ---")
        print(results[['Item_ID', 'Item_Name', 'Current_Stock']])

        # Step 4: ID Selection & Validation
        target_id = input("\nEnter the exact Item_ID to use (or 'C' to cancel): ").strip().upper()

        if target_id == 'C':
            continue # Back to search

        if target_id in df['Item_ID'].values:
            # Step 5: Professional Error Handling (Try-Except)
            try:
                current_stock = df.loc[df['Item_ID'] == target_id, 'Current_Stock'].values[0]
                is_lasa = df.loc[df['Item_ID'] == target_id, 'LASA_Warning'].values[0]

                if is_lasa:
                    print("⚠️  SAFETY ALERT: This is a LASA medication. Check twice!")

                amount_input = input(f"How many units? (Available: {current_stock}): ")
                amount = int(amount_input)

                if amount < 0:
                    print("❌ Error: Amount cannot be negative.")
                elif current_stock >= amount:
                    # Update & Save
                    df.loc[df['Item_ID'] == target_id, 'Current_Stock'] -= amount
                    df.to_csv('clinic_inventory.csv', index=False)
                    print(f"✅ Success! Remaining: {current_stock - amount}")
                else:
                    print("❌ Error: Not enough stock available!")

            except ValueError:
                print(f"❌ Input Error: '{amount_input}' is not a valid number. Please enter digits only.")
            except Exception as e:
                print(f"❌ An unexpected error occurred: {e}")
        else:
            print("❌ Invalid Item_ID.")
    else:
        print(f"❌ No items found matching '{item_to_find}'.")
```
